### Control Structure

Review the cell below what value will be printed?

```python
# create set of relevent ICBs
commissioners = {'QNQ', 'QNX', 'QRL', 'QSL', 'QU9'} 
# create set of relevent NHS Trusts
trusts = {
    'R1C', 'R1F', 'RA4', 'RBA', 'RDR', 'RDU', 'RH5', 'RHM', 'RHU', 
    'RHW', 'RN5', 'RNU', 'RPC', 'RTH', 'RW1', 'RWX', 'RX2', 'RXC',
    'RXH', 'RXQ', 'RYR'
}

code  = 'RXQ '

# Identify org_type of code
if code in commissioners:
    print('ICB')
elif code in trusts:
    print('NHS Trust')
else:
    print('Other')
```

### Repetition Structure

Review the cell below, what number will be printed?

```python
i = 1

while i < 50:
    i += 1
    i *= i

print(i)
```
